## Responsible Use of Large Language Models (LLMs)

In this lab, **you are allowed and encouraged to use LLMs responsibly** as learning tools.
Think of them as **tutors, reference books, and debugging partners** — not as answer generators.

### Appropriate uses
- Asking for **explanations** of Python concepts (lists, loops, functions, conditionals)
- Getting **hints** or alternative approaches when you are stuck
- Debugging errors *after* you try to reason about them yourself
- Asking an LLM to **explain your own code** back to you

### Not appropriate
- Copy‑pasting complete solutions without understanding them
- Submitting code you cannot explain
- Using an LLM instead of thinking through the problem first

You may be asked to explain your code or reflect briefly on how you used an LLM.

### Commonly used LLMs (examples)

- **ChatGPT** — https://chat.openai.com  
  General‑purpose reasoning, explanations, and debugging. Good for step‑by‑step thinking.

  IDE‑integrated suggestions. Treat suggestions as *ideas*, not answers.

- **Perplexity** — https://www.perplexity.ai

No single tool is required or preferred. What matters is **how** you use it.

*Exercise 3:* Modify Exercise 2 so that it takes a matrix in the format from Exercise 1 and draws a tic‑tac‑toe board with `"X"`s and `"O"`s.

````python
def draw_ttt(board):
    symbols = {0: " ", 1: "X", 2: "0"}
    n = len(board)
    m = len(board[0])
    for i in range(n):
        print(" ---" * m)
        for j in range(m):
            print(f"| {symbols[board[i][j]]} ", end="")
        print("|")
    print(" ---" * m)
````

*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
def check_status(board):
    n = len(board)
    for row in board:
        if row[0] != 0 and all(cell == row[0] for cell in row):
            return row[0]
    for col in range(n):
        first = board[0][col]
        if first != 0 and all (board[row][col] == first for row in range(n)):
            return first
    first = board[0][0]
    if first != 0 and all(board[i][i] == first for i in range(n)):
        return first
    first = board[0][n-1]
    if first !=0 and all(board[i][n-1-i] == first for i in range(n)):
        return first 
    for row in board:
        if 0 in row:
            return -1
    return 0
````

*Exercise 5:* Write a function that takes a game board, a player number, and `(row, col)` coordinates and places the correct mark (`"X"` or `"O"`) in that location.

Requirements:

- Only allow placing a mark in a previously empty location.
- Return `True` if the move was successful, and `False` otherwise.


````python
def place_move(board, player, row, col):
    n = len(board)
    if row < 0 or row >= n or col < 0 or col >= n:
        return False
    if player not in (1, 2):
        return False
    if board[row][col] != 0:
        return False
    board[row][col] = player
    return True
````

````python
board = [
    [0,0,0],
    [0,0,0],
    [0,0,0]]
print(place_move(board, 1,0,0))
print(place_move(board, 2,0,0))
print(place_move(board, 2,1,1))
print(place_move(board, 1,3,0))
````

*Exercise 6:* Modify Exercise 3 to show **row and column labels** so that players can specify locations like `"A2"` or `"C1"`.

````python
def draw_labeled_ttt(board):
    symbols = {0: " ", 1: "X", 2: "O"}
    n = len(board)
    print("    ", end="")
    for col in range(n):
        print(f"  {chr(ord('A') + col)} ", end="")
    print()
    for i in range(n):
        print("   " + " ---" * n)
        print(f"{i + 1:2} ", end="")
        for j in range(n):
            print(f"| {symbols[board[i][j]]} ", end="")
        print("|")
    print("   " + " ---" * n)
````

*Exercise 7:* Write a function that takes a board, a player number, and a location string (as in Exercise 6), then uses your function from Exercise 5 to update the board.

````python
def make_move(board, player, location):
    if len(location) < 2:
        return False  
    col_letter = location[0].upper()
    row_number = location[1:]
    col = ord(col_letter) - ord('A')
    if not row_number.isdigit():
        return False
    row = int(row_number) - 1
    return place_move(board, player, row, col)
````

````python
board = create_board(3)
print(make_move(board, 1, "A1"))
print(make_move(board, 1, "B2"))
print(make_move(board, 1, "A1"))
print(make_move(board, 1, "C3"))
print(make_move(board, 1, "D4"))
print(make_move(board, 1, "B0"))
draw_labeled_ttt(board)
````

*Exercise 8:* Write a function that is called with a board and player number, takes input from the player using Python's `input()`, and modifies the board using your function from Exercise 7.

Keep asking for input until the player enters a valid location that results in a valid move.


````python
def player_turn(board, player):
    while True:
        location = input(f"Player {player} ('{'X' if player == 1 else 'O'}'), enter your move (e.g., A1): ").strip()
        if make_move(board, player, location):
            print(f"Move accepted: {location.upper()}")
            break
        else:
            print(f"Invalid move: {location}. Try again.")
````

*Exercise 11:* Develop a version of the game where one player is the computer.

Note: you do **not** need an extensive search for the best move. For example, you can have the computer:
- block obvious losses
- otherwise try to create a winning row/column/diagonal


````python
import random
def computer_move(board, computer_player):
    opponent = 2 if computer_player == 1 else 1
    n = len(board)
    def try_move(row, col, player):
        board[row][col] = player
        result = check_status(board)
        board[row][col] = 0
        return result == player
    for i in range(n):
        for j in range(n):
            if board[i][j] == 0 and try_move(i, j, computer_player):
                board[i][j] = computer_player
                print(f"Computer ({'X' if computer_player == 1 else 'O'}) plays at {chr(ord('A')+j)}{i+1}")
                return
    for i in range(n):
        for j in range(n):
            if board[i][j] == 0 and try_move(i, j, opponent):
                board[i][j] = computer_player
                print(f"Computer ({'X' if computer_player == 1 else 'O'}) blocks at {chr(ord('A')+j)}{i+1}")
                return
    empty_cells = [(i, j) for i in range(n) for j in range(n) if board[i][j] == 0]
    if empty_cells:
        i, j = random.choice(empty_cells)
        board[i][j] = computer_player
        print(f"Computer ({'X' if computer_player == 1 else 'O'}) plays at {chr(ord('A')+j)}{i+1}")
````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

````python
def play_vs_computer(n, human_player=1):
    board = create_board(n)
    computer_player = 2 if human_player == 1 else 1
    current_player = 1  
    while True:
        draw_labeled_ttt(board)
        if current_player == human_player:
            player_turn(board, human_player)
        else:
            computer_move(board, computer_player)
        status = check_status(board)
        if status == 1:
            draw_labeled_ttt(board)
            winner = "Human" if human_player == 1 else "Computer"
            print(f"{winner} (X) wins!")
            break
        elif status == 2:
            draw_labeled_ttt(board)
            winner = "Human" if human_player == 2 else "Computer"
            print(f"{winner} (O) wins!")
            break
        elif status == 0:
            draw_labeled_ttt(board)
            print("It's a draw!")
            break
        current_player = 2 if current_player == 1 else 1
````

*Exercise 13:* Make the 2 computer players play each-other for 10 games on a 3x3, then 4x4, then 5x5 grid. Set the max depth so that the games only take seconds. Measure the "smarter" player's win rate for each grid.

````python
def simple_computer_move(board, player, smart=True):
    n = len(board)
    opponent = 2 if player == 1 else 1
    if smart:
        for i in range(n):
            for j in range(n):
                if board[i][j] == 0:
                    board[i][j] = player
                    if check_status(board) == player:
                        return
                    board[i][j] = 0
        for i in range(n):
            for j in range(n):
                if board[i][j] == 0:
                    board[i][j] = opponent
                    if check_status(board) == opponent:
                        board[i][j] = player
                        return
                    board[i][j] = 0
    empty_cells = [(i, j) for i in range(n) for j in range(n) if board[i][j] == 0]
    if empty_cells:
        i, j = random.choice(empty_cells)
        board[i][j] = player
````

````python
def simulate_simple_game(n, smart_player=1):
    board = create_board(n)
    current_player = 1
    while True:
        if current_player == smart_player:
            simple_computer_move(board, current_player, smart=True)
        else:
            simple_computer_move(board, current_player, smart=False)
        status = check_status(board)
        if status != -1:
            return status
        current_player = 2 if current_player == 1 else 1
````

````python
def benchmark_simple_ai(n, num_games=10):
    wins_smart = 0
    wins_random = 0
    draws = 0
    for _ in range(num_games):
        smart_first = random.choice([1, 2])
        winner = simulate_simple_game(n, smart_player=smart_first)
        if winner == smart_first:
            wins_smart += 1
        elif winner == 0:
            draws += 1
        else:
            wins_random += 1
    win_rate = wins_smart / num_games
    print(f"{n}x{n} Grid: Smart wins {wins_smart}, Random wins {wins_random}, Draws {draws}, Smart win rate: {win_rate:.2f}")
````

````python
benchmark_simple_ai(3, num_games=10)
benchmark_simple_ai(4, num_games=10)
benchmark_simple_ai(5, num_games=10)
````
